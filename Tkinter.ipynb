```python
# version pour un placement auto des widgets 

from tkinter import *

#creation d'un fenêtre 
fenetre = Tk()

mon_texte = Label(fenetre, text = "Essayons d'afficher un commentaire")
mon_texte.pack()

mon_texte2 = Label(fenetre,text = "Deuxième tentative")
mon_texte2.pack()

#boucle
fenetre.mainloop()
```

```python
# version pour un placement contrôlé des widgets

from tkinter import *
from tkinter import ttk

# création d'ine fenêtre 
fenetre =Tk()

# création d'un cadre dans la fenêtre et d'une grille
frm=ttk.Frame(fenetre)
frm.grid()

# création d'un texte en vue d'un placement dans le cadre 
mon_texte=ttk.Label(frm,text="Essayons d'ouvrir une fenêtre pour y afficher un message .")

# affichage du texte dans le cadre, selon les coordonnées précisées dans la grille
mon_texte.grid(column = 0,row = 0)

mon_texte2 =ttk.Label(frm,text="Deuxième texte")
mon_texte2.grid(column = 1,row = 0)

mon_texte3 =ttk.Label(frm,text="Troisème texte")
mon_texte3.grid(column = 2,row = 1)

text_user =ttk.Label(frm,text="UserName")
text_user.grid(column = 1, row =2)

saisie = Entry(frm,bd = 10, bg="green")
saisie.grid(column = 1, row = 2)

fenetre.mainloop()
```
