```python
#Calcul moyenne UE1

R101 = float(input("Note R101 : "))
R102 = float(input("Note R102 : "))
R103 = float(input("Note R103 : "))

coef_R101 = float(input("Coefficient R101 : "))
coef_R102 = float(input("Coefficient R102 : "))
coef_R103 = float(input("Coefficient R103 : "))

total = (
    R101 * coef_R101
    + R102 * coef_R102
    + R103 * coef_R103 )

coefficients = (
    coef_R101
    + coef_R102
    + coef_R103 )

UE1 = total / coefficients

print("Moyenne UE1 :", UE1, "/ 20")


```
